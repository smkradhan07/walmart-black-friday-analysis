```python
#import the libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

#Load Dataset
walmart = pd.read_csv("walmart_data.csv")
walmart.head()
```

```python
#Missing Values & Outlier Detection

#Missing value
walmart.isnull().sum()
```

```python
#Outlier Detection
mu = walmart['Purchase'].mean()
sigma = walmart['Purchase'].std()

upper = mu + 3*sigma
lower = mu - 3*sigma

outliers = walmart[(walmart['Purchase'] > upper) |
                    (walmart['Purchase'] < lower)]

print("Number of Outliers:", len(outliers))
```

```python
#Boxplot for Outliers
plt.figure(figsize=(8,5))
sns.boxplot(x=walmart['Purchase'])
plt.title("Outlier Detection")
plt.show()
```
