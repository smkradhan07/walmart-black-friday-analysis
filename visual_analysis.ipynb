```python
#import the libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

#Load Dataset
walmart = pd.read_csv("walmart_data.csv")
walmart.head()
```

```python
#Question 3:Visual Analysis

#Histogram + KDE (Purchase)
plt.figure()
sns.histplot(walmart['Purchase'], kde=True)
plt.title("Distribution of Purchase Amount")
plt.show()
```

```python
#Boxplot (Gender vs Purchase)
plt.figure()
sns.boxplot(x='Gender', y='Purchase', data=walmart)
plt.title("Purchase by Gender")
plt.show()
```

```python
#Boxplot (Age vs Purchase)
plt.figure()
sns.boxplot(x='Age', y='Purchase', data=walmart)
plt.xticks(rotation=45)
plt.title("Purchase by Age")
plt.show()
```

```python
#Correlation Heatmap
plt.figure()
corr = walmart.corr(numeric_only=True)
sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
