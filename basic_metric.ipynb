```python
#import the libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

#Load Dataset
walmart = pd.read_csv("walmart_data.csv")
walmart.head()
```

```python
#Question 1: Basic Metrics & Problem Understanding

#shape of Dataset
print("Shape of data:", walmart.shape)
```

```python
#Datatype of Attributes
print("\nData Types:\n", walmart.dtypes)
```

```python
#Conversion of categorical columns to "Category"
categorical_cols = ['Gender', 'Age', 'City_Category',
                    'Stay_In_Current_City_Years', 'Marital_Status']

for col in categorical_cols:
    walmart[col] = walmart[col].astype('category')

print(walmart.dtypes)
```
